```python
import pandas as pd
import numpy as np
```

```python
personal = pd.read_csv('/home/ililo/archive/spotify_data.csv')#personal.shape(69742, 23)
personal.columns.tolist() #columns to use: [['ts', 'ms_played', 'master_metadata_track_name', 'master_metadata_album_artist_name', 'master_metadata_album_album_name', 'reason_start', 'reason_end', 'skipped']]
personal = personal[['ts', 'ms_played', 'master_metadata_track_name', 'master_metadata_album_artist_name', 'master_metadata_album_album_name', 'reason_start', 'reason_end', 'skipped']]
personal = personal.rename(columns={'master_metadata_track_name': 'track',
                                    'master_metadata_album_artist_name': 'artist',
                                    'master_metadata_album_album_name': 'album'})
personal['artist'] = personal['artist'].astype(str).str.lower()
personal ['album'] = personal['album'].astype(str).str.lower()
personal['track'] = personal['track'].astype(str).str.lower()
personal['ts'] = pd.to_datetime(personal['ts'])
personal['id'] = personal['track'] + '_' + personal['artist']
personal.shape#shape(69742, 9)
personal.head()
```

columns to use

```python
tracks = pd.read_csv('/home/ililo/archive/tracks.csv')
tracks.columns.tolist()#columns to use: [['name', 'popularity', 'duration_ms', 'artists', 'danceability', 'energy', 'key', 'loudness', 'mode', 'speechiness', acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo', 'time_signature']]
tracks = tracks[['name', 'popularity', 'duration_ms', 'artists', 'danceability', 'energy', 'key', 'loudness', 'mode', 'speechiness', 'acousticness', 'instrumentalness', 'liveness', 'valence', 'tempo', 'time_signature']]
tracks = tracks.rename(columns={'name': 'track',
                                'artists': 'artist'})
tracks[['artist', 'track']] = tracks[['artist', 'track']].apply(lambda x: x.astype(str).str.lower().str.strip().str.replace(r"[\[\]\'\-\"(){}]", "", regex=True))
tracks['id'] = tracks['track'] + '_' + tracks['artist']
tracks.shape#shape(586672, 17)
tracks.head()
```

```python
merged = pd.merge(personal, tracks, on=['id'], how='inner')
merged.shape#shape(22727, 25)
merged.drop(columns=['track_y', 'artist_y'], inplace=True)
merged.rename(columns={'track_x': 'track',
                       'artist_x': 'artist'}, inplace=True)
merged.to_csv('merged.csv', index=False)
merged.head(50)
```
