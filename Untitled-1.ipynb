```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/home/ililo/Dev/merged.csv')
df['ts'] = pd.to_datetime(df['ts']).dt.tz_convert('Africa/Addis_Ababa')
df = df.sort_values('ts')
df
```

```python
split_point = int(len(df) * 0.8)

# Split without shuffling
train_df = df.iloc[:split_point].copy()
test_df  = df.iloc[split_point:].copy()
```

```python
def create_time_features(df):
    df = df.copy()

    df['hour'] = df['ts'].dt.hour
    df['day_of_week'] = df['ts'].dt.dayofweek
    df['month'] = df['ts'].dt.month
    df['is_weekend'] = df['day_of_week'].isin([5, 6]).astype(bool)

    df['time_of_day'] = pd.cut(df['hour'], 
                            bins=[0, 4, 12, 17, 24],
                            labels=['Mid-Night', 'Morning', 'Afternoon', 'Night'],
                            include_lowest=True)
    return df
```

```python
def create_completion_features(df):
    df = df.copy()
    df['rewound'] = (df['ms_played'] > df['duration_ms']).astype(bool)
    df['completion_rate'] = (df['ms_played']/df['duration_ms'])*100

    return df
```

```python
def create_categorical_features(df):

    df=df.copy()
    
    mapping = {
        'clickrow': 'new',
        'fwdbtn': 'user',
        'trackdone': 'natural',
        'backbtn': 'user',
        'endplay': 'natural'}

    df['reason_start'] = df['reason_start'].map(mapping)
    df['reason_end'] = df['reason_end'].map(mapping)

    return df
```

```python
def create_engagement_features(df, reference_stats=None):

    df = df.copy()
    
    if reference_stats is None:

        track_stats= df.groupby('id').agg(
            avg_completion_rate=('completion_rate', 'mean'),
            streams = ('id', 'count')
        ).reset_index()
    else:

        track_stats=reference_stats

    df = pd.merge(df, track_stats, on='id', how='left')
    return df, track_stats
```

```python
def apply_features(train_df, test_df=None):
    # Enhance: Add informative docstring
    """
    Applies feature creation pipelines to train and (optionally) test DataFrames. 
    Also aligns aggregation stats for consistency between data splits.
    
    Args:
        train_df (pd.DataFrame): Training dataset.
        test_df (pd.DataFrame, optional): Testing dataset. Defaults to None.
    
    Returns:
        train_df (pd.DataFrame): Training dataset with new features.
        test_df (pd.DataFrame): Testing dataset with new features (if given).
    """

    # Process training data
    train_df = create_time_features(train_df)
    train_df = create_categorical_features(train_df)
    train_df = create_completion_features(train_df)
    train_df, track_stats = create_engagement_features(train_df, reference_stats=None)

    # Process test data if provided
    if test_df is not None:
        test_df = create_time_features(test_df)
        test_df = create_categorical_features(test_df)
        test_df = create_completion_features(test_df)
        # Enhance: Consistent engagement features statistics usage
        test_df, _ = create_engagement_features(test_df, reference_stats=track_stats)

        # Enhance: Remove potential duplicates before re-merging track_stats
        test_df = test_df.drop(['avg_completion_rate', 'streams'], axis=1, errors='ignore')
        test_df = pd.merge(test_df, track_stats, on='id', how='left', suffixes=('', '_stat'))

        # Enhance: Fill missing completion and streams using global statistics
        global_avg = train_df['completion_rate'].mean()
        test_df['completion_rate'] = test_df['completion_rate'].fillna(global_avg)
        test_df['streams'] = test_df['streams'].fillna(0)

        # Enhance: Ensure the returned test_df has a deterministic column order, if desired
        # test_df = test_df.loc[:, train_df.columns]  # Uncomment if strict column alignment is needed

    return train_df, test_df
```

```python
def calculate_preference_score(df, weights=(0.6, 0.7)):
 
    df['streams'] = df.groupby('id')['id'].transform('count')

    df['norm_completion'] = df['avg_completion_rate'] / 100
    df['norm_streams'] = np.log1p(df['streams']) / np.log1p(df['streams'].max())

    df['preference_score'] = (
        weights[0] * df['norm_completion'] +
        weights[1] * df['norm_streams']
    ) * 100

    df['preference_score'] = df['preference_score'].clip(0, 100)
    
    return df
```

```python
train_df, reference_stats = apply_features(train_df)
train_df.head(50)
```

```python
train_df = calculate_preference_score(train_df)
test_df, reference_stats = apply_features(test_df)
test_df.columns.tolist()
```

```python
track_streams = train_df.groupby('id', as_index=False).agg({'streams': 'count'})
tot_streams = track_streams.sort_values('streams', ascending=False).head(50)
tot_streams
```

```python
target = 'preference_score'
num_features = ['ms_played', 'popularity', 'duration_ms', 'danceability', 'energy', 'key', 'loudness', 'mode', 'speechiness', 'acousticness', 'instrumentalness', 'valence', 'tempo', 'liveness', 'time_signature', 'hour', 'day_of_week', 'month', 'completion_rate', 'avg_completion_rate', 'streams']
cat_features = ['reason_start', 'reason_end', 'skipped', 'is_weekend', 'time_of_day', 'rewound']
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_features)
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', RandomForestRegressor
        (n_estimators=200, random_state=2, n_jobs=-1
        ))
])

X_train = train_df.drop(columns=[target])
y_train = train_df[target]

X_test = test_df.copy()
#y_test = test.df[target]

model.fit(X_train, y_train)
preds = model.predict(X_test)
```

```python
split_idx = int(len(train_df) * 0.8)
train_fold = train_df.iloc[:split_idx]
val_fold = train_df.iloc[split_idx:]

A_train = train_fold.drop(columns=[target])
b_train = train_fold[target]

A_val = val_fold.drop(columns=[target])
b_val = val_fold[target]

model.fit(A_train, b_train)
val_preds = model.predict(A_val)

from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(b_val, val_preds)
r2 = r2_score(b_val, val_preds)

print(f"Validation MSE {mse: .4f}, R2: {r2: .4f}")
```
